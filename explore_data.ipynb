```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
file_path = "brainTumorDataPublic_1-766/1.mat"

with h5py.File(file_path, 'r') as f:
    cjdata = f['cjdata']
    
    label = int(np.array(cjdata['label']).item())
    image = np.array(cjdata['image'])
    tumor_mask = np.array(cjdata['tumorMask'])
    pid_raw = np.array(cjdata['PID'])
    
    print("Label:", label)
    print("Image shape:", image.shape)
    print("Tumor mask shape:", tumor_mask.shape)
    print("PID raw:", pid_raw)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image, cmap='gray')
axes[0].set_title("MRI Image")
axes[1].imshow(tumor_mask, cmap='gray')
axes[1].set_title("Tumor Mask")
plt.show()
```

```python
pid_string = ''.join(chr(int(c[0])) for c in pid_raw)
print("Patient ID (as string):", pid_string)
```
